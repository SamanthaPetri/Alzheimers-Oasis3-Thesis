```python
import nibabel as nib
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('D:/mamba_model/thesis_cohort_final.csv')

subj = df.iloc[0]
pet_v2 = nib.load(f"E:/Oasis3/PET_registered_v2/{subj['subject_id']}_PIB_in_MRI.nii.gz").get_fdata()
mri_raw = nib.load(f"E:/Oasis3/FastSurfer_output/{subj['mri_session']}/mri/orig.nii.gz").get_fdata()

print(f"PET shape: {pet_v2.shape}")

fig, axes = plt.subplots(2, 3, figsize=(15, 10))

mid_x = pet_v2.shape[0] // 2
mid_y = pet_v2.shape[1] // 2
mid_z = pet_v2.shape[2] // 2

axes[0, 0].imshow(mri_raw[mid_x, :, :], cmap='gray', origin='lower')
axes[0, 0].set_title('MRI Sagittal')
axes[0, 1].imshow(mri_raw[:, mid_y, :], cmap='gray', origin='lower')
axes[0, 1].set_title('MRI Coronal')
axes[0, 2].imshow(mri_raw[:, :, mid_z], cmap='gray', origin='lower')
axes[0, 2].set_title('MRI Axial')

axes[1, 0].imshow(pet_v2[mid_x, :, :], cmap='hot', origin='lower')
axes[1, 0].set_title('PET Sagittal')
axes[1, 1].imshow(pet_v2[:, mid_y, :], cmap='hot', origin='lower')
axes[1, 1].set_title('PET Coronal')
axes[1, 2].imshow(pet_v2[:, :, mid_z], cmap='hot', origin='lower')
axes[1, 2].set_title('PET Axial')

plt.tight_layout()
plt.show()
```
